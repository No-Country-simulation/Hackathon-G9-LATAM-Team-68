1.- Definir prototipo de datos entrantes

```python
usuario = {
    "ingreso_mensual": 15000,
    "ahorro_mensual": 1500,
    "pago_deudas_mensual": 2000,

    "transacciones": [
        {
            "descripcion": "Supermercado Walmart",
            "valor": 2500
        },
        {
            "descripcion": "Gasolina Pemex",
            "valor": 1200
        },
        {
            "descripcion": "Netflix",
            "valor": 219
        }
    ]
}
```

```python
print(usuario)
```

2.- Clasificacion de transacciones

```python
def clasificar_transaccion(descripcion):

    descripcion = descripcion.lower()

    if "walmart" in descripcion:
        return "alimentacion"

    elif "supermercado" in descripcion:
        return "alimentacion"

    elif "gasolina" in descripcion:
        return "transporte"

    elif "pemex" in descripcion:
        return "transporte"

    elif "netflix" in descripcion:
        return "ocio"

    else:
        return "otros"
```

```python
categoria = clasificar_transaccion(
    "Supermercado Walmart"
)

print(categoria)
```

```python
for transaccion in usuario["transacciones"]:

    descripcion = transaccion["descripcion"]

    categoria = clasificar_transaccion(
        descripcion
    )

    print(
        descripcion,
        "→",
        categoria
    )
```

3.- Agregar la categoria a cada transaccion 

```python
for transaccion in usuario["transacciones"]:

    categoria = clasificar_transaccion(
        transaccion["descripcion"]
    )

    transaccion["categoria"] = categoria
```

```python
print(usuario["transacciones"])
```

4.- Agrupar gastos

```python
resumen_gastos = {}
```

```python
for transaccion in usuario["transacciones"]:

    categoria = transaccion["categoria"]
    valor = transaccion["valor"]

    if categoria in resumen_gastos:

        resumen_gastos[categoria] += valor

    else:

        resumen_gastos[categoria] = valor
```

```python
print(resumen_gastos)
```

5.- Calcular indicadores financieros

```python
gasto_total = sum(
    transaccion["valor"]
    for transaccion in usuario["transacciones"]
)
tasa_ahorro = (
    usuario["ahorro_mensual"]
    / usuario["ingreso_mensual"]
)
nivel_endeudamiento = (
    usuario["pago_deudas_mensual"]
    / usuario["ingreso_mensual"]
)
tasa_gasto = (
    gasto_total
    / usuario["ingreso_mensual"]
)
indicadores = {
    "tasa_ahorro": tasa_ahorro,
    "nivel_endeudamiento": nivel_endeudamiento,
    "tasa_gasto": tasa_gasto
}
```

```python
print(indicadores)
```

6.- Crear perfil financiero

```python
def clasificar_perfil(
    tasa_ahorro,
    nivel_endeudamiento,
    tasa_gasto
):

    if (
        tasa_ahorro >= 0.10
        and nivel_endeudamiento <= 0.30
        and tasa_gasto <= 0.80
    ):
        return "Saludable"

    elif (
        nivel_endeudamiento <= 0.50
        and tasa_gasto <= 1.00
    ):
        return "En observacion"

    else:
        return "En riesgo"
```

```python
perfil = clasificar_perfil(
    tasa_ahorro,
    nivel_endeudamiento,
    tasa_gasto
)

print(perfil)
```

7.- Generar recomendaciones

```python
recomendaciones = []
```

```python
if tasa_ahorro < 0.10:

    recomendaciones.append(
        "Considera aumentar gradualmente tu ahorro mensual."
    )


if nivel_endeudamiento > 0.30:

    recomendaciones.append(
        "Revisa tu nivel de endeudamiento."
    )


if tasa_gasto > 0.80:

    recomendaciones.append(
        "Tus gastos representan una parte elevada de tus ingresos."
    )
```

```python
if len(recomendaciones) == 0:

    recomendaciones.append(
        "Mantén tus hábitos financieros actuales."
    )
```

8.- Construir resultado final

```python
resultado = {
    "perfil_financiero": perfil,

    "resumen_gastos": resumen_gastos,

    "indicadores": indicadores,

    "recomendaciones": recomendaciones
}
```

```python
print(resultado)
```

Convertir a JSON

```python
import json
resultado_json = json.dumps(
    resultado,
    indent=4,
    ensure_ascii=False
)
print(resultado_json)
```
